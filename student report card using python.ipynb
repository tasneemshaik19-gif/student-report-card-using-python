```python
from PIL import Image, ImageDraw, ImageFont

# Function to calculate grade
def calculate_grade(percentage):
    if percentage >= 90:
        return "A+"
    elif percentage >= 80:
        return "A"
    elif percentage >= 70:
        return "B"
    elif percentage >= 60:
        return "C"
    elif percentage >= 50:
        return "D"
    else:
        return "F"

# Input
name = input("Enter Student Name: ")
roll = input("Enter Roll Number: ")

marks = []
for i in range(1, 6):
    mark = float(input(f"Enter Subject {i} Marks: "))
    marks.append(mark)

# Calculations
total = sum(marks)
percentage = total / 5
grade = calculate_grade(percentage)
result = "PASS" if percentage >= 35 else "FAIL"

# Create Image
img = Image.new("RGB", (700, 500), color="white")
draw = ImageDraw.Draw(img)

font_title = ImageFont.load_default()
font = ImageFont.load_default()

draw.text((180, 20), "STUDENT REPORT CARD", fill="blue", font=font_title)

details = [
    f"Student Name : {name}",
    f"Roll Number  : {roll}",
    "",
    f"Subject 1 : {marks[0]}",
    f"Subject 2 : {marks[1]}",
    f"Subject 3 : {marks[2]}",
    f"Subject 4 : {marks[3]}",
    f"Subject 5 : {marks[4]}",
    "",
    f"Total Marks : {total}/500",
    f"Percentage  : {percentage:.2f}%",
    f"Grade       : {grade}",
    f"Result      : {result}"
]

y = 70
for line in details:
    draw.text((50, y), line, fill="black", font=font)
    y += 30

# Save image
img.save("student_result.png")

print("\nResult image saved as 'student_result.png'")
```
